### 3.2 Breadboard
### Understanding the Breadboard

A breadboard is used to build circuits without soldering. It allows you to easily connect components like LEDs, resistors, and wires.

The breadboard is divided into three main sections:

1. **Middle Section (Terminal Strips):**
   - The central area is split into two halves by a vertical gap.
   - The left side has rows labeled **A to E**, and the right side has rows labeled **F to J**.
   - Each row of 5 holes (A–E) is internally connected horizontally.
   - Each row of 5 holes (F–J) is also internally connected horizontally.
   - However, **there is NO connection between E and F across the gap**.

2. **Power Rails (Top and Bottom):**
   - The long rows marked with **+ (positive)** and **– (negative)** run along the top and bottom.
   - These are connected **vertically in columns**, used to distribute power (VCC and GND) across the board.

### Important Notes for This Experiment

- If you place two component legs in the **same row (A–E or F–J)**, they are electrically connected.
- If you place them across **E and F**, they are **NOT connected**, because of the central gap.
- This gap is useful when placing components like an LED or IC, so their legs are not shorted together.
- Always ensure that:
  - The **resistor is in series with the LED** (not in the same row).
  - The **LED legs are not placed in the same connected row**, otherwise it will not work properly.

This structure helps you safely and clearly build circuits like the LED blinking experiment.

### 3.4 Jumper Wires
Jumper wires are used to make temporary electrical connections between components like the Raspberry Pi, breadboard, LEDs, and resistors. They help you quickly build and modify circuits without soldering.

For your reference, an image is provided below showing three types of jumper wires:

- The **top red wire** is a **male-to-male** jumper wire (both ends have pins).
- The **middle yellow wire** is a **male-to-female** jumper wire (one pin end and one socket end).
- The **bottom orange wire** is a **female-to-female** jumper wire (both ends have sockets).

You can identify them based on their ends:
- **Male ends** are used to plug into the breadboard.
- **Female ends** are used to connect to GPIO pins on the Raspberry Pi.

**Note:** In most of our experiments, we will mainly use **male-to-female jumper wires** to connect the Raspberry Pi GPIO pins to the breadboard.

### Understanding Raspberry Pi GPIO Pins

The Raspberry Pi provides a set of GPIO (General Purpose Input/Output) pins that allow us to connect and control external components such as LEDs, sensors, and other devices.

For this experiment, we use the Raspberry Pi Zero 2W. To correctly identify the pins, hold the Raspberry Pi in the following orientation:

- USB ports on the **left side**
- GPIO pins on the **right side**
- SD card at the **top**
- Camera connector at the **bottom**

In this orientation, the GPIO pins are arranged in two vertical columns:

- Left column: odd-numbered pins (1, 3, 5, ...)
- Right column: even-numbered pins (2, 4, 6, ...)

You can count the pins from top to bottom.

## 6. Wiring Steps

### Wiring Explanation (Refer to the Image Below)

Refer to the wiring image below for this example setup.

In this circuit:

- The **long leg of the LED** is placed in **F40**.
- The **short leg of the LED** is placed in **E40** (across the center gap, so they are not directly connected).

The connections are made as follows:

- The **long leg (F40)** is connected to a **jumper wire** placed in the same row (for example H40), and this wire is connected to **GPIO17 (Pin 11)** on the Raspberry Pi.
- The **short leg (E40)** is connected to one end of the **resistor**, which is placed in **D40**. Since E40 and D40 are in the same row, they are electrically connected.
- The **other end of the resistor** is placed in **D35**.
- A **black jumper wire** is connected in the same row (A35), which is connected to **D35**.
- This black wire then goes to a **GND pin (Pin 6)** on the Raspberry Pi.

### Key Idea

The resistor is connected **in series** with the LED, meaning the current flows through:

GPIO → LED → Resistor → GND

This ensures safe operation of the LED and the Raspberry Pi.

## 8. Important Preparation Note

### Important Note on LED Orientation

The LED has two legs:

- **Long leg → positive (anode)**
- **Short leg → negative (cathode)**

In this circuit:
- The long leg must be connected toward the GPIO pin
- The short leg must go toward the resistor and then to ground

If the LED is reversed, it will not light up.

### Common Mistakes to Avoid

- Placing both LED legs in the **same breadboard row** (this will not work)
- Connecting the LED **without a resistor** (can damage the LED or Raspberry Pi)
- Reversing LED polarity (LED will not glow)
- Connecting across the breadboard gap incorrectly (no connection between E and F rows)
- Loose jumper wire connections

Always double-check your wiring before running the code.

## 10. LED Blink Test

## 11. From LED Control to Human Perception

In the previous steps, we controlled the LED in a simple way: ON or OFF.

However, in real life, many things are not binary. For example, brightness is not just "on" or "off". We describe it using words like:

- very dark  
- dark  
- medium  
- bright  
- very bright  

These are called **linguistic terms**.

The meaning of these terms is not exact. It depends on **human perception** and **context**.

This idea is the basis of **fuzzy logic**, where values are not strictly true or false, but can belong partially to different categories.

## 12. Conceptual Example

Consider the following situation:

A person's height is **172 cm**.

How would you describe this height?

- very short  
- short  
- medium  
- tall  
- very tall  

The answer depends on the person:

- In some regions, 172 cm may be considered **tall**
- In other regions, it may be considered **average or even short**

This shows that human reasoning is not always precise.  
It is based on perception rather than fixed numerical boundaries.

In this experiment, we will use this idea to control an LED.

## 13. Experiment 1: Perception-Based LED Blinking

Now I ask you to rate the height (172 cm) using a simple scale:

1 → very short  
2 → short  
3 → medium  
4 → tall  
5 → very tall  

Based on this input, the LED will blink the same number of times.

This shows how a **subjective human perception** can control a physical system.

```python
led.close()
print("Previous LED object released")
```

```python
from gpiozero import LED
from time import sleep

led = LED(17)

print("Rate the height 172 cm:")
print("1 = very short")
print("2 = short")
print("3 = medium")
print("4 = tall")
print("5 = very tall")

rating = int(input("Enter your rating (1–5): "))

for i in range(rating):
    led.on()
    sleep(0.5)
    led.off()
    sleep(0.5)

print(f"LED blinked {rating} times based on your perception.")
```

## 14. Connecting to Fuzzy Logic

In this experiment:

- A **numerical value** (input from user)  
- is interpreted using **linguistic categories**  
- and then mapped to a **physical output (LED)**  

This is a simplified version of a fuzzy logic system.

Instead of strict boundaries, we use **gradual interpretation** based on perception.

This approach is useful in many real-world applications such as:

- smart lighting systems  
- temperature control  
- decision support systems
